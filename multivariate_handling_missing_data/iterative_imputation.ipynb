```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
df = pd.read_csv("./iterative_imputation_practice.csv")
df.sample(5)
```

```python
df.isnull().sum()
```

```python
X = df.drop(columns=["Salary"])
y = df["Salary"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
imputer = IterativeImputer(max_iter=5, random_state=2)
```

```python
X_train_impute = imputer.fit_transform(X_train)
X_test_impute = imputer.transform(X_test)
```

```python
pd.DataFrame(X_train_impute, columns=X_train.columns)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
lr = LinearRegression()

lr.fit(X_train_impute, y_train)

y_pred = lr.predict(X_test_impute)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("R2 Score:", r2)
```
